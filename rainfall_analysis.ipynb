This notebook cleans rainfall data, performs NumPy analysis and generates planting recommendations.

```python
import numpy as np
import csv

counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70],
]

machakos_raw = [
    "48", "35", "88", "N/A", "140",
    " 20 ", "9", "12", "",
    "45", "182", "96"
]

months = [
    "January","February","March","April",
    "May","June","July","August",
    "September","October","November","December"
]
```

```python
def clean_machakos(data):
    valid_values = []

    for value in data:
        try:
            valid_values.append(int(value.strip()))
        except ValueError:
            pass

    replacement = round(sum(valid_values) / len(valid_values))

    cleaned = []
    imputed_months = []

    for month, value in zip(months, data):
        try:
            cleaned.append(int(value.strip()))
        except ValueError:
            cleaned.append(replacement)
            imputed_months.append(month)

    print("Imputed months:", imputed_months)
    print("Replacement value:", replacement)

    return cleaned

machakos_clean = clean_machakos(machakos_raw)
print(machakos_clean)
```

```python
all_data = clean_rows + [machakos_clean]

R = np.array(all_data)

print("Shape:", R.shape)
print("Data type:", R.dtype)
```

```python
annual_totals = R.sum(axis=1)

print("Annual Totals")
for county, total in zip(counties, annual_totals):
    print(county, total)
```

```python
monthly_avg = R.mean(axis=0)

print("Monthly Average Rainfall")
for month, avg in zip(months, monthly_avg):
    print(month, round(avg, 2))
```

```python
wettest_indices = np.argmax(R, axis=1)

wettest_months = [months[i] for i in wettest_indices]

print("Wettest Month Per County")

for county, month in zip(counties, wettest_months):
    print(county, "-", month)
```

```python
dry_months = np.sum(R < 50, axis=1)

print("Dry Month Counts")

for county, count in zip(counties, dry_months):
    print(county, count)
```

```python
two_month_totals = R[:, :-1] + R[:, 1:]

best_window_idx = np.argmax(two_month_totals, axis=1)

planting_windows = [
    f"{months[i]}-{months[i+1]}"
    for i in best_window_idx
]

print("Recommended Planting Windows")

for county, window in zip(counties, planting_windows):
    print(county, "-", window)
```

```python
with open("rainfall_summary.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow([
        "county",
        "annual_total_mm",
        "wettest_month",
        "dry_months",
        "planting_window"
    ])

    for county, total, wettest, dry, window in zip(
        counties,
        annual_totals,
        wettest_months,
        dry_months,
        planting_windows
    ):
        writer.writerow([
            county,
            total,
            wettest,
            dry,
            window
        ])

print("rainfall_summary.csv created successfully")
```

# Planting Advisory

Rainfall patterns indicate that the most favorable planting periods occur during the consecutive two‑month windows with the highest combined rainfall. Nairobi and Kisumu receive substantial rainfall during April and May, making this period suitable for crop establishment. Nakuru also experiences its strongest rainfall during April and May, supporting rain‑fed agriculture. Mombasa receives its heaviest rainfall during May and June, while Machakos benefits most from the November–December season. Farmers should align planting schedules with these peak rainfall windows to maximize germination and early crop growth.

The Machakos station data contained invalid entries in April and September. These values were replaced using the rounded average of valid Machakos observations. While this approach preserves the overall rainfall pattern and enables analysis, the affected months may not reflect actual field conditions.
